```python
import os
import pandas as pd
from pathlib import Path

# ── 1. تعريف المسارات الرسمية في كاجل ──────────────────────────────────────────
# مسار الداتا سيت الأصلية (الصور)
BASE_IMG_DIR = "/kaggle/input/datasets/divg07/casia-20-image-tampering-detection-dataset/CASIA2"
# مسار الميزات التي استخرجتها (تأكد من مطابقة اسم المجلد في الـ Input)
FEATURES_DIR = "/kaggle/input/datasets/abdulrahmnessawu/processed-features-data/processed_features"
# مسار الأقنعة (Ground Truth)
MASK_DIR = "/kaggle/input/datasets/divg07/casia-20-image-tampering-detection-dataset/CASIA2/CASIA 2 Groundtruth"

def create_final_csv():
    data_list = []
    tp_folder = Path(BASE_IMG_DIR) / "Tp"
    
    print("🔍 جاري فحص الملفات وربطها... برجاء الانتظار.")
    
    # التأكد من وجود المجلدات أولاً
    if not tp_folder.exists():
        print(f"❌ خطأ: لم يتم العثور على مجلد الصور في {tp_folder}")
        return
    
    # المرور على جميع الصور المزورة (Tp)
    for img_path in tp_folder.glob("*"):
        if img_path.suffix.lower() not in ['.jpg', '.png', '.tif', '.jpeg']:
            continue
            
        stem = img_path.stem
        
        # إنشاء المسارات المتوقعة للميزات والقناع
        mask_path = Path(MASK_DIR) / f"{stem}_gt.png"
        ela_path = Path(FEATURES_DIR) / f"{stem}_ela.npy"
        noise_path = Path(FEATURES_DIR) / f"{stem}_noise.npy"
        edge_path = Path(FEATURES_DIR) / f"{stem}_edge.npy"
        
        # التحقق من وجود "الخماسي" (الصورة + القناع + 3 ميزات)
        if all([mask_path.exists(), ela_path.exists(), noise_path.exists(), edge_path.exists()]):
            data_list.append({
                'image_id': stem,
                'image_path': str(img_path),
                'mask_path': str(mask_path),
                'ela_path': str(ela_path),
                'noise_path': str(noise_path),
                'edge_path': str(edge_path)
            })

    # تحويل القائمة لجدول DataFrame
    df = pd.DataFrame(data_list)
    
    if not df.empty:
        # ── 2. حفظ ملف الـ CSV في مخرجات النوت بوك ──────────────────────────────
        output_file = "/kaggle/working/final_manifest.csv"
        df.to_csv(output_file, index=False)
        print("-" * 50)
        print(f"✅ مبروك! تم الربط بنجاح.")
        print(f"📊 عدد العينات السليمة للتدريب: {len(df)}")
        print(f"💾 المسار الملموس للملف: {output_file}")
        print("-" * 50)
    else:
        print("⚠️ لم يتم العثور على ملفات متطابقة. تأكد من إضافة الـ Dataset لـ Input النوت بوك.")

# تشغيل المهمة
create_final_csv()
```
